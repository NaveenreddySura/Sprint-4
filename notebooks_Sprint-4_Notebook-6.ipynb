# Multivariate Analysis

## Introduction to Multivariate Analysis

Multivariate analysis is a data analysis technique used to study relationships among three or more variables at the same time. It helps us understand how multiple factors work together and influence patterns in a dataset.

## Correlation Matrix
Definition
A correlation matrix is a table that shows the correlation coefficient between every pair of numerical variables in a dataset. Its values range from -1 to +1, indicating the direction and strength of a linear relationship.
### Real-world example
- Study how height, weight, age, and BMI are related to each other.
- Analyze the relationships among study hours, attendance, practice tests, and exam marks.
### Business example
- Compare sales, advertising expenses, product prices, and customer visits to understand how these business metrics relate.
- Analyze revenue, profit, marketing costs, and order volume to identify related performance indicators.
### AI/ML use case
A correlation matrix helps identify redundant numerical features and potential multicollinearity before training linear regression and other statistical models.

```python
%mamba install pandas
```

```python
import pandas as pd

df = pd.DataFrame({
    "Advertising": [10, 20, 30, 40, 50],
    "Sales": [20, 40, 60, 80, 100],
    "Profit": [5, 10, 15, 20, 25],
    "Returns": [50, 40, 30, 20, 10]
})

corr_matrix = df.corr().round(2)
print(corr_matrix)
```

### Explanation of the code
- df.corr() calculates correlations between all numerical columns.
- A coefficient of +1.0 indicates a perfect positive linear relationship.
- A coefficient of -1.0 indicates a perfect negative linear relationship.
- A coefficient near 0 indicates little or no linear relationship.
- This dataset is deliberately constructed with perfect relationships for learning; real business data is rarely this exact.

## Heatmap
Definition
A heatmap is a visual representation of numerical values using different colors. In multivariate analysis, a correlation heatmap uses colors to show the strength and direction of relationships between numerical variables.
### Real-world example
- Visualize correlations between height, weight, age, and BMI.
- Identify which study habits have the strongest relationships with students' exam marks.
### Business example
- A company uses a heatmap to compare correlations among sales, profit, advertising expenses, and discounts.
- An e-commerce business visualizes relationships between customer visits, order value, delivery time, and customer ratings.
### AI/ML use case
Heatmaps help data scientists quickly identify highly correlated features, possible multicollinearity, and relationships that deserve further investigation before model training.

```python
%mamba install seaborn
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Advertising": [10, 20, 30, 40, 50],
    "Sales": [20, 40, 60, 80, 100],
    "Profit": [5, 10, 15, 20, 25],
    "Returns": [50, 40, 30, 20, 10]
})

corr = df.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.show()
```

### Explanation of the code
- df.corr() calculates the correlation matrix.
- sns.heatmap() converts the matrix into a color-coded chart.
- annot=True displays correlation values inside the cells.
- cmap="coolwarm" sets a color gradient for positive and negative values.
- vmin=-1 and vmax=1 keep the scale consistent with the correlation range.
- plt.show() displays the chart.

## Pair Plot
Definition
A pair plot displays pairwise relationships between multiple numerical variables in a single figure. It typically uses scatter plots for relationships between different variables and distribution plots along the diagonal.
### Real-world example
- Compare height, weight, age, and BMI to identify relationships and unusual observations.
- Explore how study hours, attendance, practice tests, and marks relate to one another.
### Business example
- Analyze sales, profit, discount, and advertising expenses together to discover business patterns.
- Compare customer age, annual income, spending score, and purchase value to understand customer segments.
### AI/ML use case
Pair plots are useful during Exploratory Data Analysis (EDA) to identify feature relationships, clusters, possible outliers, and patterns that may help distinguish classification groups.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Age": [22, 25, 28, 32, 35, 40],
    "Income": [25000, 30000, 35000, 45000, 50000, 65000],
    "Spending": [80, 75, 70, 60, 55, 45]
})

sns.pairplot(df)
plt.show()
```

### Explanation of the code
- The DataFrame contains three numerical features.
- sns.pairplot(df) generates pairwise plots for the selected columns.
- Scatter plots show relationships between different variables.
- Diagonal plots show the distribution of each individual variable.
- The charts help identify trends, clusters, and possible outliers without creating each plot separately.

## Multiple Feature Relationships
Definition
Multiple feature relationship analysis examines how several independent variables are related to one another and to a target variable. It helps us understand a target more comprehensively than analyzing one feature at a time.
### Real-world example
- A student's exam marks may be associated with study hours, attendance, sleep, and practice tests.
- A house's price may be related to its area, number of bedrooms, age, and location.
### Business example
- A business examines how advertising cost, discounts, and website visitors together relate to sales.
- A retailer analyzes price, season, stock availability, and marketing campaigns to understand product demand.
### AI/ML use case
Multiple feature analysis helps determine which combination of inputs may be useful for predicting a target variable. It supports feature selection, regression modeling, and model interpretation.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    "Area": [800, 1000, 1200, 1500, 1800],
    "Bedrooms": [1, 2, 2, 3, 3],
    "Age": [20, 15, 10, 8, 5],
    "Price": [20, 28, 35, 45, 55]
})

X = df[["Area", "Bedrooms", "Age"]]
y = df["Price"]

model = LinearRegression()
model.fit(X, y)

print("Predicted price:",
      round(model.predict([[1400, 3, 7]])[0], 2))
```

### Explanation of the code
- X contains three input features: area, bedrooms, and age.
- y contains the target variable, house price.
- LinearRegression() creates a multiple linear regression model.
- fit(X, y) learns coefficients from all three features together.
- predict() estimates the price for a house with an area of 1,400 square feet, three bedrooms, and an age of seven years.
- This is a demonstration, not a reliable real-estate pricing model. The sample is too small for dependable predictions.

## Feature Interactions
Definition
A feature interaction occurs when the relationship between one feature and a target variable depends on the value of another feature. In other words, the combined effect of two features can differ from the sum of their separate effects.
### Real-world example
- Exercise may improve performance differently depending on how much sleep a person gets.
- The effect of studying for an extra hour may differ between students with low attendance and students with high attendance.
### Business example
- A discount may increase sales more during a festival than during an ordinary week.
- Advertising may generate more sales when a product is also competitively priced.
### AI/ML use case
Feature interactions help Machine Learning models capture combined effects that individual features alone may not explain. They are particularly useful in regression, decision trees, gradient boosting, and feature engineering.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    "Advertising": [1, 2, 3, 4, 5, 6],
    "Discount": [1, 2, 1, 2, 1, 2],
    "Sales": [12, 24, 32, 48, 52, 72]
})

df["Interaction"] = (
    df["Advertising"] * df["Discount"]
)

X = df[["Advertising", "Discount", "Interaction"]]
y = df["Sales"]

model = LinearRegression()
model.fit(X, y)

print(df[["Advertising", "Discount", "Interaction"]])
print("Interaction coefficient:",
      round(model.coef_[2], 2))
```

### Explanation of the code
- Advertising and Discount are the original features.
- The new Interaction feature is created by multiplying advertising by discount.
- X includes both original features and their interaction.
- LinearRegression() estimates how each feature contributes to the target, holding the other included features constant.
- The interaction coefficient indicates the additional effect associated with the product of the two features.
- The example is constructed so the interaction coefficient is approximately 1.0.

## Numerical Feature Relationships
Definition
Numerical feature relationship analysis examines how numerical input variables relate to each other and to a numerical target. It can reveal trends, redundancy, and patterns that may be useful for prediction.
### Real-world example
- Analyze how height and weight are related across a group of people.
- Study the relationship between temperature, rainfall, and crop yield.
### Business example
- Examine how product price, advertising spending, and customer visits relate to revenue.
- Analyze how delivery distance, delivery time, and shipping cost vary together.
### AI/ML use case
Numerical feature analysis supports feature selection, regression modeling, scaling decisions, and multicollinearity checks. It helps determine which features might provide useful information for predicting a target.

```python
import pandas as pd

df = pd.DataFrame({
    "Area": [500, 700, 900, 1100, 1300],
    "Bedrooms": [1, 1, 2, 2, 3],
    "Price": [15, 20, 28, 32, 40]
})

print("Correlation with Price:")
print(df.corr()["Price"].round(2))
```

### Explanation of the code
- df.corr() calculates pairwise correlations for numerical columns.
- ["Price"] selects the correlations with the target variable, Price.
- round(2) displays values to two decimal places.
- Area and Bedrooms both have strong positive correlations with Price in this sample.
- This does not establish that these features independently cause higher prices; the features may also be related to one another.

## Identifying Important Relationships
These are the key patterns you should learn to recognize in a correlation matrix or heatmap.

### A. Strong correlations
Definition
A strong correlation indicates that two variables have a strong linear relationship. The coefficient is close to +1 or -1.
### Real-world example
- Height and weight may show a positive relationship in some populations.
- Distance traveled and fuel consumed may show a positive relationship under similar driving conditions.
### Business example
- Advertising spending and sales may increase together.
- Units sold and total revenue may be strongly related when prices are relatively stable.
### AI/ML use case
Strong correlations can help identify potentially useful features. However, if two input features are highly correlated with each other, they may provide redundant information.

```python
import pandas as pd

df = pd.DataFrame({
    "Advertising": [10, 20, 30, 40, 50],
    "Sales": [20, 40, 60, 80, 100]
})

print(round(df["Advertising"].corr(df["Sales"]), 2))
```

### Explanation of the code
- corr() calculates the Pearson correlation between Advertising and Sales.
- The result 1.0 represents a perfect positive linear relationship in the example.
- In real data, a strong correlation is not proof that advertising alone caused the increase in sales.

### B. Weak correlations
Definition
A weak correlation indicates that two variables have little linear association, with the coefficient closer to zero. They may still have a nonlinear relationship or a relationship involving other variables.
### Real-world example
- Shoe size may have little linear relationship with exam scores in a group of students.
- A person's birth month may have little linear relationship with their daily travel distance.
### Business example
- Employee ID numbers generally have no meaningful relationship with sales performance.
- A randomly assigned coupon code may have little relationship with customer spending.
### AI/ML use case
Weakly correlated features may be candidates for removal, but correlation alone should not determine feature importance. A feature with weak pairwise correlation can still help a nonlinear model or interact with other features.

```python
import pandas as pd

df = pd.DataFrame({
    "Feature": [1, 2, 3, 4, 5],
    "Target": [10, 12, 9, 11, 10]
})

print(round(df["Feature"].corr(df["Target"]), 2))
```

### Explanation of the code
- corr() measures the linear relationship between Feature and Target.
- A result of 0.0 indicates no linear correlation in this example.
- It does not prove that no relationship of any kind exists.

### C. Negative correlations
Definition
A negative correlation means that higher values of one variable tend to be associated with lower values of another variable. A coefficient close to -1 represents a strong negative linear relationship.
### Real-world example
- When travel speed increases over a fixed distance, travel time tends to decrease.
- As the price of a product rises, the quantity demanded may decrease, depending on market conditions.
### Business example
- Higher product prices may be associated with lower sales volume.
- Longer delivery times may be associated with lower customer satisfaction.
### AI/ML use case
Negative correlations help models and analysts identify inverse relationships. They can be useful when interpreting regression coefficients and exploring potential predictors.

```python
import pandas as pd

df = pd.DataFrame({
    "Price": [10, 20, 30, 40, 50],
    "Quantity": [100, 80, 60, 40, 20]
})

print(round(df["Price"].corr(df["Quantity"]), 2))
```

### Explanation of the code
- Price increases while Quantity decreases.
- corr() calculates the Pearson correlation coefficient.
- The result -1.0 indicates a perfect negative linear relationship in this constructed dataset.
- Real markets are more complicated, so this exact relationship should not be assumed in practice.

### D. Potential multicollinearity
Definition
Multicollinearity occurs when two or more input features in a model are strongly linearly related to one another. It can make individual regression coefficients difficult to interpret and less stable, even when the model's predictions are reasonably accurate.
### Real-world example
- A dataset may contain both height in centimeters and height in meters, which represent the same measurement.
- A dataset may include both a person's weight and a BMI value calculated using their weight and height.
### Business example
- A sales model may include advertising spending and a marketing budget total that already includes that spending.
- A house-price model may include total area and several area measurements that are strongly related.
### AI/ML use case
Detecting multicollinearity is important when building linear and logistic regression models. Depending on the goal, data scientists may remove redundant features, combine them, or use regularization methods such as Ridge regression.

```python
import pandas as pd

df = pd.DataFrame({
    "Area": [500, 600, 700, 800, 900],
    "Area_Squared": [1000, 1200, 1400, 1600, 1800],
    "Price": [15, 18, 21, 24, 27]
})

corr = df.corr()

print(corr.round(2))
print("\nArea correlation:",
      round(df["Area"].corr(df["Area_Squared"]), 2))
```

### Explanation of the code
- Area and Area_Squared are perfectly linearly related in this example because the second feature is twice the first.
- df.corr() calculates correlations between every pair of numerical columns.
- The coefficient of 1.0 between these features indicates perfect linear dependence.
- This is a clear example of potential multicollinearity.
- In real projects, you can also use the Variance Inflation Factor (VIF) to investigate multicollinearity among multiple features.

### E. Interesting feature relationships
Definition
Interesting feature relationships are patterns among multiple variables that reveal useful information about a dataset. These patterns may include clusters, unexpected trends, nonlinear relationships, or combined effects.
### Real-world example
- Students who sleep adequately and study regularly may perform differently from students who study for the same number of hours but sleep less.
- People with similar incomes may have different spending habits depending on age and lifestyle.
### Business example
- Discounts may increase sales more for certain product categories than for others.
- Customers who visit a website frequently and receive personalized offers may purchase more than customers who receive generic offers.
### AI/ML use case
Discovering interesting relationships can help with feature engineering, customer segmentation, anomaly detection, and model improvement. Analysts can use scatter plots, pair plots, domain knowledge, and interaction features to investigate such patterns.

```python
import pandas as pd

df = pd.DataFrame({
    "Visits": [1, 2, 3, 4, 5, 6],
    "Discount": [0, 10, 0, 10, 0, 10],
    "Sales": [10, 25, 30, 55, 50, 85]
})

df["Visit_Discount"] = df["Visits"] * df["Discount"]

print(df)
print("\nCorrelations:")
print(df.corr()["Sales"].round(2))
```

### Explanation of the code
- Visits records how frequently customers visit.
- Discount records the discount offered.
- Visit_Discount is an interaction feature created by multiplying Visits by Discount.
- df.corr()["Sales"] calculates the correlation of each numerical feature with Sales.
- These results help identify potentially useful relationships, but a high correlation alone does not prove that a feature causes higher sales.
