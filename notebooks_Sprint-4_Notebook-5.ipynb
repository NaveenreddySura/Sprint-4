# What is Bivariate Analysis

Definition: Bivariate analysis is a data analysis technique used to study the relationship between two variables. It helps us understand whether changes in one variable are associated with changes in another variable.

# Types of Bivariate Analysis

## Numerical vs Numerical

Numerical vs Numerical analysis examines the relationship between two variables that contain numerical values. It helps identify whether the variables move in the same direction, opposite directions, or have no clear relationship.
### Real-world example
- Study the relationship between a person's age and their annual income.
- Analyze whether spending more time exercising is associated with burning more calories.
### Business example
- A company studies advertising expenditure and sales revenue to understand how marketing investment relates to sales.
- A retailer compares product prices with the number of units sold to identify possible pricing patterns.
### AI/ML use case
- Identifying relationships between numerical features before training a regression model.
- Detecting highly correlated features that may create redundancy in some Machine Learning models.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Age": [22, 25, 28, 32, 35, 40],
    "Salary": [25000, 30000, 35000, 45000, 50000, 65000]
})

sns.scatterplot(data=df, x="Age", y="Salary")
plt.title("Age vs Salary")
plt.show()

print("Correlation:", round(df["Age"].corr(df["Salary"]), 2))
```

### Explanation of the 
- DataFrame() creates a dataset containing age and salary.
- sns.scatterplot() displays each person's age and salary as a point.
- plt.title() adds a title to the chart.
- plt.show() displays the scatter plot.
- corr() calculates the correlation coefficient. The result of 0.99 indicates a very strong positive linear relationship in this sample.

## Numerical vs Categorical
Definition
Numerical vs Categorical analysis compares numerical values across different categories or groups. It helps us understand how a numerical variable's distribution changes between categories.
### Real-world example
- Compare the monthly salaries of employees working in different departments.
- Compare the exam marks of students from different courses.
### Business example
- An organization compares salaries across IT, HR, and Finance departments.
- A retail company compares the average sales revenue of different product categories.
### AI/ML use case
- Identifying whether a categorical feature, such as department or product type, is useful for predicting a numerical target.
- Understanding differences between groups before training regression models or performing feature engineering.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Salary": [50000, 60000, 35000, 40000, 45000, 55000]
})

print(df.groupby("Department")["Salary"].mean())
```

### Explanation of the code
- Department is the categorical variable, and Salary is the numerical variable.
- groupby("Department") groups employees by department.
- ["Salary"].mean() calculates the average salary within each group.
- The result shows that IT has the highest average salary in this sample.

## Categorical vs Categorical
Definition
Categorical vs Categorical analysis studies the relationship between two variables that contain labels or groups rather than numerical measurements. It helps identify patterns in how categories occur together.
### Real-world example
- Analyze the relationship between gender categories and preferred shopping methods.
- Study whether a person's education level is associated with their employment status.
### Business example
- A company compares department and employee status, such as Active or Resigned.
- An online store examines the relationship between product category and payment method.
### AI/ML use case
- Finding patterns between categorical features before building classification models.
- Identifying potentially useful features for predicting customer churn, loan approval, or purchase behavior.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT", "HR"],
    "Status": ["Active", "Resigned", "Active",
               "Active", "Active", "Resigned"]
})

table = pd.crosstab(df["Department"], df["Status"])
print(table)
```

### Explanation of the code
- Both Department and Status are categorical variables.
- pd.crosstab() counts how frequently each combination occurs.
- The table shows two Active employees and one Resigned employee in each department.
- This is a count-based comparison; proportions can also be calculated to compare groups of different sizes.

# Bivariate Analysis Techniques

## Scatter Plot
Definition
A scatter plot is a graph that displays the relationship between two numerical variables using individual points. Each point represents one observation, making it easier to identify trends, clusters, and unusual values.
### Real-world example
- Plot study hours on the X-axis and exam marks on the Y-axis.
- Plot daily temperature against electricity consumption to see whether hotter days are associated with higher energy usage.
### Business example
- Plot advertising expenditure against sales revenue to examine whether increased marketing investment is associated with higher sales.
- Plot product price against quantity sold to investigate customer purchasing patterns.
### AI/ML use case
Scatter plots help analysts explore feature relationships, detect outliers, and assess whether a linear regression model may be appropriate.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [35, 45, 55, 70, 85]
})

plt.scatter(df["Hours"], df["Marks"])
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

### Explanation of the code
- plt.scatter() creates the scatter plot.
- Hours is plotted on the X-axis.
- Marks is plotted on the Y-axis.
- xlabel() and ylabel() label the axes.
- The upward trend indicates a positive relationship in this dataset.

## Box Plot
Definition
A box plot summarizes the distribution of numerical data using the median, quartiles, and whiskers. It helps compare the spread of numerical values across categories and identify potential outliers.
### Real-world example
- Compare the salary distributions of employees across different departments.
- Compare students' marks across different classes to identify differences in typical scores and variability.
### Business example
- Compare delivery times across different courier companies.
- Compare the order values of customers across membership tiers to identify differences in spending patterns.
### AI/ML use case
Box plots help detect potential outliers, compare feature distributions between classes, and identify data-quality issues before model training.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Department": ["IT"] * 4 + ["HR"] * 4,
    "Salary": [40000, 45000, 50000, 90000,
               30000, 32000, 35000, 38000]
})

sns.boxplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.show()
```

### Explanation of the code
- The dataset contains salaries for two departments.
- sns.boxplot() creates a separate box for each department.
- The line inside each box represents the median.
- The box represents the middle 50% of the observations, between the first and third quartiles.
- Whiskers extend to values within the box-plot rule, and individual points beyond them may be potential outliers.
- A potential outlier is not automatically an error; it should be investigated.

## Violin Plot
Definition
A violin plot shows the distribution of a numerical variable across categories, combining a box-plot summary with a density shape. The width of the violin indicates where observations are more concentrated.
### Real-world example
- Compare the distribution of exam marks across different classrooms.
- Compare the heights of people in different age groups to understand where values are concentrated.
### Business example
- Compare customer spending distributions across Bronze, Silver, and Gold membership levels.
- Compare delivery times across delivery partners to understand typical times and variation.
### AI/ML use case
Violin plots help identify differences in data distributions, multiple concentration regions, and variation across classes before selecting features for a Machine Learning model.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Department": ["IT"] * 5 + ["HR"] * 5,
    "Salary": [40000, 45000, 50000, 55000, 60000,
               30000, 32000, 35000, 36000, 38000]
})

sns.violinplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.show()
```

### Explanation of the code
- Department divides the observations into IT and HR groups.
- Salary contains the numerical measurements.
- sns.violinplot() displays the distribution shape for each department.
- Wider portions indicate salary values with higher estimated density.
- Unlike a box plot, a violin plot reveals more detail about the distribution's shape, although the shape is estimated from the available data.

## Grouped Bar Chart
Definition
A grouped bar chart displays multiple sets of bars side by side for each category. It is useful for comparing numerical counts or aggregated values across two categorical variables.
### Real-world example
- Compare the number of male and female students enrolled in different courses.
- Compare the number of customers using cash, card, or online payments across different stores.
### Business example
- Compare sales of different product categories across multiple branches.
- Compare the numbers of new and returning customers across different marketing campaigns.
### AI/ML use case
Grouped bar charts help explore relationships between categorical features and identify differences in class frequencies, customer segments, and purchasing behavior.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Category": ["Food", "Food", "Clothing", "Clothing"],
    "Payment": ["Cash", "Online", "Cash", "Online"],
    "Sales": [100, 150, 120, 180]
})

pivot = df.pivot(
    index="Category",
    columns="Payment",
    values="Sales"
)

pivot.plot(kind="bar")
plt.title("Sales by Category and Payment Method")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()

print(pivot)
```

### Explanation of the code
- pivot() arranges the data into a table with product categories as rows and payment methods as columns.
- plot(kind="bar") creates a grouped bar chart.
- Each category contains separate bars for Cash and Online sales.
- ylabel() labels the vertical axis.
- The chart makes it easy to compare sales across both dimensions.
### Note: 
This example uses one value for each category-payment combination. With multiple transactions, aggregate sales first using groupby() or a pivot table with aggfunc="sum".

## Cross Tabulation
Definition
Cross tabulation is a technique that creates a frequency table showing how observations are distributed across combinations of two categorical variables. It can display counts, row percentages, or column percentages.
### Real-world example
- Count how many people from different age groups prefer different payment methods.
- Compare survey responses such as Yes and No across different education levels.
### Business example
- Count how many customers in each region purchase each product category.
- Analyze customer churn by subscription plan to identify groups with higher cancellation rates.
### AI/ML use case
Cross tabulation helps investigate categorical feature relationships, examine target-class distributions, and discover patterns relevant to classification models.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Food", "Food", "Food", "Clothing",
                "Clothing", "Clothing"],
    "Payment": ["Cash", "Online", "Online",
                "Cash", "Cash", "Online"]
})

table = pd.crosstab(df["Product"], df["Payment"])
print(table)

print("\nRow percentages:")
print(pd.crosstab(
    df["Product"],
    df["Payment"],
    normalize="index"
).round(2))
```

### Explanation of the code
- pd.crosstab() counts observations for each combination of Product and Payment.
- normalize="index" converts the counts into proportions within each product category.
- Clothing purchases are approximately 67% Cash and 33% Online in this sample.
- Food purchases are approximately 33% Cash and 67% Online.
- Percentages help compare categories fairly when their total counts differ.
