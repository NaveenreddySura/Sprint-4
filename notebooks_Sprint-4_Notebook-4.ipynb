# Univariate Analysis

Univariate analysis means analyzing one variable at a time to understand its distribution, typical values, variation, and unusual observations. Numerical variables are analyzed using statistical measures and plots, while categorical variables are analyzed using frequencies, counts, and category distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

salaries = pd.Series([20, 25, 25, 30, 35, 40, 45, 80])
```

## Mean
Definition: The mean is the arithmetic average of all numerical values in a dataset. It is calculated by adding all the values and dividing their sum by the total number of observations.

### Real-world example: 
A company calculates the average monthly salary of its employees to understand its overall salary level.

### AI/ML use case: 
The mean is used to summarize numerical features and can be used to fill missing numerical values when the data distribution makes it appropriate.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

print("Mean:", np.mean(salaries))
```

#### Explanation
- np.array() creates a NumPy array containing the salary values.
- np.mean() adds the values and divides their sum by 8.
- The average salary is ₹37,500.
- The high value of 80 increases the mean, showing that the mean can be influenced by unusually large values.

## Median
Definition: The median is the middle value after numerical data has been arranged in ascending order. If the dataset contains an even number of observations, the median is the average of the two middle values.

### Real-world example: 
A company uses the median salary to describe a typical employee's salary without letting a few very high salaries dominate the result.

### AI/ML use case: 
The median is often used to summarize skewed numerical features and to impute missing values when extreme values may distort the mean.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

print("Median:", np.median(salaries))
```

### Explanation
- The sorted values are already in ascending order.
- There are 8 values, so the two middle values are 30 and 35.
- NumPy averages these values: \((30 + 35) / 2 = 32.5\).
- The median salary is ₹32,500 and is less affected by the high salary of 80 than the mean is.

## Mode
Definition: The mode is the value that occurs most frequently in a dataset. A dataset can have one mode, multiple modes, or no uniquely most frequent value.

### Real-world example: 
A retailer finds the most commonly purchased shoe size to decide which sizes to stock more frequently.

### AI/ML use case: 
The mode is commonly used to summarize categorical variables and to fill missing categorical values when appropriate.

```python
import pandas as pd

salaries = pd.Series([20, 25, 25, 30, 35, 40, 45, 80])

print("Mode:")
print(salaries.mode().tolist())
```

### Explanation
- pd.Series() stores the values in a Pandas Series.
- .mode() returns the value or values with the highest frequency.
- The value 25 occurs twice, while each other value occurs once.
- Therefore, 25 (₹25,000) is the mode.

## Minimum
Definition: The minimum is the smallest value in a numerical dataset. It helps identify the lower boundary of the observed values.

### Real-world example: 
A business checks the lowest monthly salary paid to an employee.

### AI/ML use case: 
The minimum helps understand feature ranges and can reveal values that require validation.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

print("Minimum salary:", np.min(salaries))
```

### Explanation
- np.min() searches for the smallest value in the array.
- The minimum is 20, representing ₹20,000.
- This is the smallest observed salary, not necessarily the minimum salary allowed by company policy.

## Maximum
Definition: The maximum is the largest value in a numerical dataset. It helps identify the upper boundary of the observed values.

### Real-world example: 
A company identifies its highest monthly salary to understand the upper end of its salary distribution.

### AI/ML use case: 
The maximum can reveal extreme observations and help check whether a feature contains possible data-entry errors.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

print("Maximum salary:", np.max(salaries))
```

### Explanation
- np.max() returns the largest value.
- The maximum salary is 80, representing ₹80,000.
- A maximum value is not automatically an error; it should be checked against the business context.

## Range
Definition: The range is the difference between the maximum and minimum values in a dataset. It gives a simple measure of the total spread between the two extremes.

### Real-world example: 
A school calculates the difference between the highest and lowest exam scores.

### AI/ML use case: 
The range helps identify feature scales and spot values that may need investigation or scaling.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

salary_range = np.max(salaries) - np.min(salaries)

print("Range:", salary_range)
```

### Explanation
- np.max() returns 80 and np.min() returns 20.
- Subtracting the minimum from the maximum gives \(80 - 20 = 60\).
- The salary range is ₹60,000.
- Because it depends only on the two extremes, the range can be strongly affected by outliers.

## Variance
Definition: Variance measures how far numerical values spread around their mean by calculating the average squared deviation from that mean. A larger variance indicates greater spread, and variance is expressed in squared units.

### Real-world example: 
Two classes have similar average marks, but one class has much more variation in students' marks.

### AI/ML use case: 
Variance helps identify features with little variation and supports understanding how numerical data is distributed.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

print("Variance:", np.var(salaries))
```

### Explanation
- np.var() calculates the average squared distance of the values from their mean.
- NumPy uses population variance by default, dividing by the number of observations.
- The variance is 322.1875 in squared salary units (thousands of rupees squared).
- The high salary value of 80 contributes to the spread.

## Standard Deviation
Definition: Standard deviation measures how spread out numerical values are around their mean. It is the square root of the variance, so it is expressed in the same units as the original data.

### Real-world example: 
A manager compares salary variation across two departments, even when their average salaries are similar.

### AI/ML use case: 
Standard deviation helps assess feature spread, standardize numerical features, and identify unusually distant values.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

print("Standard deviation:", round(np.std(salaries), 2))
```

### Explanation
- np.std() calculates the population standard deviation by default.
- The result is approximately 17.95, in thousands of rupees.
- A larger standard deviation means the observations are more spread out around their mean.
- Standard deviation alone does not tell you whether a distribution is skewed or contains multiple groups.

## Quartiles
Definition: Quartiles divide sorted numerical data into four parts, each representing approximately 25% of the observations. The first quartile (Q1), second quartile (Q2 or median), and third quartile (Q3) describe the distribution's lower, middle, and upper portions.

### Real-world example: 
A company examines the lower 25%, median, and upper 25% of salaries to understand employee compensation.

### AI/ML use case: 
Quartiles help detect potential outliers using the interquartile range and summarize feature distributions.

```python
import numpy as np

salaries = np.array([20, 25, 25, 30, 35, 40, 45, 80])

q1 = np.percentile(salaries, 25)
q2 = np.percentile(salaries, 50)
q3 = np.percentile(salaries, 75)

print("Q1:", q1)
print("Q2 (Median):", q2)
print("Q3:", q3)
```

### Explanation
- np.percentile() calculates the value at the specified percentile.
- Q1 is 25, meaning approximately 25% of the distribution lies at or below that point.
- Q2 is 32.5, which is the median.
- Q3 is 41.25, representing the 75th percentile.
- Exact quartile values can differ between software methods because percentile interpolation conventions vary.

## Percentiles
Definition: A percentile indicates the value below which a specified percentage of observations falls. For example, the 90th percentile is a value at or below which approximately 90% of the observations lie.

### Real-world example: 
A school finds the 90th percentile of exam scores to understand how top-performing students compare with the rest.

### AI/ML use case: 
Percentiles help set thresholds, analyze feature distributions, and cap extreme values when a justified preprocessing strategy calls for it.

```python
import numpy as np

marks = np.array([40, 50, 55, 60, 65, 70, 80, 95])

p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p90 = np.percentile(marks, 90)

print("25th percentile:", p25)
print("50th percentile:", p50)
print("90th percentile:", p90)
```

### Explanation
- np.percentile(marks, 25) calculates the 25th percentile.
- The 50th percentile is the median.
- The 90th percentile is 84.5 using NumPy's default percentile method.
- Percentiles are useful for understanding where observations sit within the distribution, but they do not reveal the entire distribution on their own.

## Skewness
Definition: Skewness measures the asymmetry of a numerical distribution around its center. Positive skew indicates a longer right tail, negative skew indicates a longer left tail, and values near zero suggest little asymmetry.

### Real-world example: 
Most employees earn moderate salaries, while a small number of executives earn very high salaries, creating a right-skewed salary distribution.

### AI/ML use case: 
Skewness helps identify highly asymmetric features that may benefit from a transformation, depending on the model and task

```python
import pandas as pd

salaries = pd.Series([20, 22, 24, 25, 27, 30, 35, 100])

print("Skewness:", round(salaries.skew(), 2))
```

### Explanation
- .skew() calculates the sample skewness of the Series.
- The large salary value of 100 pulls the distribution's right tail outward.
- The positive result indicates right skew.
- Skewness is a clue about distribution shape, not proof that a value is an error or must be removed.

## Kurtosis
Definition: Kurtosis describes aspects of a distribution's tail heaviness and its tendency to produce extreme observations compared with a normal distribution. Pandas' .kurt() reports excess kurtosis, for which a normal distribution has a value of approximately zero.

### Real-world example: 
A financial analyst checks whether daily returns contain more extreme movements than would be expected under a normal distribution.

### AI/ML use case: 
Kurtosis helps assess tail behavior and investigate features that may contain unusually extreme observations.

```python
import pandas as pd

values = pd.Series([10, 12, 12, 13, 14, 15, 16, 40])

print("Excess kurtosis:", round(values.kurt(), 2))
```

### Explanation
- .kurt() calculates excess kurtosis using the sample-based Fisher definition.
- The value 40 is far from the other observations and contributes to the heavy-tail indication.
- Positive excess kurtosis indicates heavier tails or more extreme observations relative to a normal distribution under this measure.
- Kurtosis should be interpreted alongside plots and other statistics; it is not simply a measure of how tall a graph looks.

# Numerical Visualizations

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Salary": [20, 25, 25, 30, 35, 40, 45, 80]
})
```

## Histogram
Definition: A histogram displays the distribution of numerical data by dividing values into intervals called bins and showing how many observations fall into each bin. It helps reveal concentration, spread, skewness, gaps, and possible unusual values.

### Real-world example:
A company plots employee salaries to see which salary ranges contain the most employees.

### AI/ML use case: 
Inspect numerical features before modeling to identify skewness, multiple clusters, or unusual observations.

```python
import matplotlib.pyplot as plt

plt.hist(df["Salary"], bins=5, edgecolor="black")

plt.title("Distribution of Employee Salaries")
plt.xlabel("Salary (thousands of rupees)")
plt.ylabel("Frequency")
plt.show()
```

### Explanation
- plt.hist() divides salary values into five bins.
- edgecolor="black" outlines the bars so they are easier to distinguish.
- The x-axis represents salary ranges and the y-axis represents the count in each range.
- The histogram helps show where salaries are concentrated and whether the high value of 80 lies far from most observations.

## KDE / Density Plot
Definition: A Kernel Density Estimate (KDE) is a smooth estimate of a numerical variable's probability density. Unlike a histogram, which uses bins, a KDE uses a smoothing method to display the overall shape of the distribution.

### Real-world example: 
An analyst compares the distribution of customer spending without relying on fixed histogram bin boundaries.

### AI/ML use case: 
Explore continuous feature distributions and identify skewness or multiple peaks before selecting preprocessing methods.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.kdeplot(
    data=df,
    x="Salary",
    fill=True
)

plt.title("Salary Density Plot")
plt.xlabel("Salary (thousands of rupees)")
plt.ylabel("Estimated density")
plt.show()
```

### Explanation
- sns.kdeplot() estimates the density across the salary values.
- fill=True shades the area beneath the curve.
- The x-axis represents salary and the y-axis represents estimated density, not the number of employees.
- The curve's shape depends on the bandwidth and the small sample size, so it should not be treated as definitive evidence of the true population distribution.

## Box Plot
Definition: A box plot summarizes a numerical distribution using its quartiles, median, and whiskers. It can also highlight possible outliers based on a common rule involving the interquartile range.

### Real-world example: 
An HR analyst compares employee salaries and identifies salaries that are unusually high or low relative to the rest.

### AI/ML use case: 
Screen numerical features for possible outliers and compare spread before deciding whether special preprocessing is needed.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(data=df, x="Salary")

plt.title("Employee Salary Box Plot")
plt.xlabel("Salary (thousands of rupees)")
plt.show()
```

### Explanation
- sns.boxplot() displays the salary distribution using its quartiles.
- The box represents the middle 50% of the data.
- The line inside the box represents the median.
- Whiskers extend to the most extreme observations within the plotting rule, and separate points can indicate possible outliers.
- An outlier is not automatically incorrect; it may represent a genuine observation.

# Categorical Variables

```python
import pandas as pd

departments = pd.Series([
    "IT", "HR", "IT", "Finance",
    "HR", "IT", "Sales", "Finance"
])
```

## Frequency
Definition: Frequency is the number of times a particular value or category appears in a dataset. Examining frequencies shows which categories occur more often than others.

### Real-world example: 
A company counts how many employees work in IT, HR, Finance, and Sales.

### AI/ML use case: 
Understand class distributions and identify rare categories before encoding categorical features or training a model.

```python
import pandas as pd

departments = pd.Series([
    "IT", "HR", "IT", "Finance",
    "HR", "IT", "Sales", "Finance"
])

print("Total records:", len(departments))
print("Frequency of IT:", (departments == "IT").sum())
```

### Explanation
- len(departments) returns the total number of observations.
- departments == "IT" creates a Boolean Series identifying IT records.
- .sum() counts the True values.
- The dataset contains 8 records, of which 3 belong to IT.

## Unique values
Definition: Unique values are the distinct values present in a column, counted without repetition. They help identify the range of categories a variable contains.

### Real-world example: 
A retailer checks which product categories exist in its inventory dataset.

### AI/ML use case: 
Identify the categories that need encoding and detect unexpected labels or high-cardinality features.

```python
import pandas as pd

departments = pd.Series([
    "IT", "HR", "IT", "Finance",
    "HR", "IT", "Sales", "Finance"
])

print("Unique departments:")
print(departments.unique())

print("Number of unique departments:", departments.nunique())
```

### Explanation
- .unique() returns the distinct department values in order of first appearance.
- .nunique() counts the distinct non-missing values by default.
- The dataset has four unique departments even though it contains eight records.
- This helps verify whether the categories match the expected business labels.

## Value counts
Definition: Value counts calculate how often each distinct category appears in a column. Pandas' value_counts() sorts the counts in descending order by default.

### Real-world example: 
A marketing team counts how many customers belong to each subscription plan.

### AI/ML use case: 
Examine class imbalance in a classification target or identify dominant and rare categorical values.

```python
import pandas as pd

departments = pd.Series([
    "IT", "HR", "IT", "Finance",
    "HR", "IT", "Sales", "Finance"
])

print(departments.value_counts())
```

### Explanation
- .value_counts() counts each distinct category.
- The results are sorted from the most frequent to the least frequent category.
- IT appears 3 times, HR and Finance appear twice each, and Sales appears once.
- By default, missing values are excluded; use value_counts(dropna=False) if you also want to count missing values.

## Percentage distribution
Definition: Percentage distribution expresses each category's frequency as a percentage of all included observations. It allows comparisons between categories even when datasets have different total sizes.

### Real-world example: 
A business reports that 40% of its customers use a particular subscription plan.

### AI/ML use case: 
Understand categorical feature distributions and measure class imbalance before selecting a modeling strategy.

```python
import pandas as pd

departments = pd.Series([
    "IT", "HR", "IT", "Finance",
    "HR", "IT", "Sales", "Finance"
])

percentage = (
    departments.value_counts(normalize=True) * 100
)

print(percentage.round(2))
```

### Explanation
- value_counts(normalize=True) calculates each category's share of the non-missing observations.
- Multiplying by 100 converts proportions into percentages.
- .round(2) rounds the displayed percentages to two decimal places.
- IT represents 37.5% of the eight records, while Sales represents 12.5%.

# Categorical Visualizations

## Bar Chart
Definition: A bar chart represents category values using rectangular bars whose heights or lengths correspond to their counts or another numerical measure. It is useful for comparing frequencies across distinct categories.

### Real-world example: 
A manager compares the number of employees in each department.

### AI/ML use case: 
Visualize categorical distributions to detect dominant categories, rare categories, and possible data imbalance.

```python
import matplotlib.pyplot as plt

counts = departments.value_counts()

plt.bar(counts.index, counts.values)

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

### Explanation
- value_counts() calculates the frequency of each department.
- plt.bar() draws one bar per category.
- The x-axis represents departments, while the y-axis represents the number of employees.
- Comparing the heights makes the most and least common departments easy to identify.

## Count Plot
Definition: A count plot displays the number of observations in each categorical group. In Seaborn, countplot() performs the category counting and plotting in a single step.

### Real-world example: 
A store visualizes how many orders belong to each payment method.

### AI/ML use case: 
Inspect category frequency and spot imbalanced groups before preprocessing categorical features.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "IT", "Sales", "Finance"
    ]
})

sns.countplot(data=df, x="Department")

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

### Explanation
- sns.countplot() counts the observations in each department automatically.
- data=df identifies the DataFrame, and x="Department" specifies the categorical column.
- The y-axis displays the number of observations per category.
- Unlike the previous bar chart, you do not need to calculate value_counts() separately.

## Pie Chart
Definition: A pie chart displays categories as slices of a circle, with each slice representing a proportion of the total. It is most appropriate when the categories are mutually exclusive and their percentages add up to the whole.

### Real-world example: 
A business displays the percentage of orders paid using cash, cards, and UPI.

### AI/ML use case: 
Present categorical proportions to stakeholders during exploratory data analysis, although bar charts are usually more useful for detailed comparisons.

```python
import matplotlib.pyplot as plt

department_counts = departments.value_counts()

plt.pie(
    department_counts.values,
    labels=department_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Department Distribution")
plt.show()
```

### Explanation
- value_counts() calculates the number of records in each department.
- plt.pie() converts the counts into slices proportional to the total.
- labels names each slice, and autopct="%1.1f%%" displays percentages to one decimal place.
- startangle=90 rotates the starting position of the chart.
- Pie charts become harder to interpret when there are many categories or percentages are similar; in those cases, prefer a bar chart.
